## Human-in-the-Loop Workflow: AI-Drafted Promotion Narratives

### 1. Scenario

An HR platform uses a generative model to draft **promotion-readiness narratives** from a manager's brief and performance evidence. Unreviewed AI output is risky here because narratives directly influence career-affecting decisions: the model can hallucinate achievements, apply inconsistent confidence language for equivalent performance, or introduce gendered framing (as measured in Module 4). An unreviewed biased narrative creates unfair outcomes and legal exposure.

### 2. Review points

- **Primary checkpoint — direct manager:** verifies factual accuracy against the evidence, checks tone/strength consistency with the performance pattern, and screens for biased framing. This is placed *before* any content reaches the promotion committee because that is the last point where errors are cheap to fix.
- **Escalation reviewer — HR Business Partner:** handles rejected drafts, suspected systematic bias, and disputes; can approve, demand regeneration, or finally reject. Separation of duties: the escalation reviewer is not the author of the brief.

### 6. Written explanation

**Why the workflow is designed this way.** The AI only ever produces a *draft*; the decision authority stays with named humans. A single mandatory checkpoint (direct manager) keeps review overhead proportionate, while the escalation path adds senior scrutiny exactly where risk concentrates (rejections, suspected bias, disputes) instead of slowing every narrative.

**How review decisions are made.** The reviewer applies four explicit criteria — factual accuracy against evidence, framing consistency with the performance pattern, absence of biased language, and appropriate confidence strength. *Approve* releases the narrative; *Edit* distinguishes immaterial wording fixes (proceed) from material changes to the rating or recommendation (regenerate and re-review, so humans never quietly launder a different recommendation through an edit); *Reject* blocks the content and forces escalation so failure modes are never silently discarded; *Escalate* routes ambiguous or sensitive cases upward.

**How escalation prevents unsafe outcomes.** Every rejected draft reaches the HR Business Partner with a logged rationale, which (a) stops managers from regenerating until something slips through, and (b) gives one role visibility across teams to spot *systematic* model problems rather than one-off errors. Final rejection is recorded and feeds the regression suite.

**How human oversight improves trust and reliability.** Accountability is traceable — each decision carries a named reviewer, timestamp, and draft version in an immutable log, so any narrative in front of the committee can be traced to who approved it and on what basis. The feedback loop turns review effort into system improvement: edit patterns and rejection categories update prompts, bias lexicons, and guardrails, so the model's drafts converge toward needing fewer corrections while the checkpoint stays in place. Oversight is designed as collaboration with guardrails, not over-reliance in either direction: the AI accelerates writing; humans own judgment.
